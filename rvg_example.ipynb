```python
def apply_radiometric_correction(E_v, E_g, E_sn, N_n, M, N, dead_time_ns=3.2, bin_height_m=0.1):
    """
    PART 2: Sequential forward dead-time correction loop.
    Reconstructs the true, physically real photon counts from observed event profiles.
    
    Parameters:
    - dead_time_ns: Hardware dead-time duration (ATLAS PMTs are typically ~3.2 ns).
    """
    num_bins = len(E_sn)
    
    # Determine how many historical bins span the instrument dead-time window (J_d)
    delta_t = (2 * bin_height_m) / 3e8
    J_v = int(np.ceil(dead_time_ns / (delta_t * 1e9))) # e.g., 3.2ns / 0.67ns ≈ 5 bins
    
    # Initialize containers for reconstructed true parameters
    N_s = np.zeros(num_bins)   # Reconstructed true signal counts
    N_sn = np.zeros(num_bins)  # Reconstructed true total counts (Signal + Noise)
    gamma = np.zeros(num_bins) # Dynamic correction coefficient matrix
    
    # 1. Forward-running chronological dead-time correction loop
    for i in range(num_bins):
        # Calculate event detection probability per detector per pulse
        P_sn_i = E_sn[i] / (M * N)
        
        # Edge-case check: prevent division-by-zero or math errors if highly saturated
        if P_sn_i >= 1.0:
            P_sn_i = 0.999  
            
        # Look backwards in time to grab the history of true photons that fired inside the dead-time window
        start_j = max(0, i - J_v)
        sum_prev_N_sn = np.sum(N_sn[start_j:i])
        
        # 2. Apply Equation 2 from Zhang et al. (2024)
        exponent_term = np.exp(sum_prev_N_sn / M)
        log_argument = 1.0 - (P_sn_i * exponent_term)
        
        # Numerical protection: handle complete detector paralyzation limits safely
        if log_argument <= 0.0:
            log_argument = 1e-5 
            
        # Reconstruct true signal photon count for this bin
        N_s[i] = -M * np.log(log_argument) - N_n[i]
        N_s[i] = max(0.0, N_s[i]) # True physical photon count cannot be below zero
        
        # Log total count (signal + noise) to feed into subsequent loops' dead-time lookback
        N_sn[i] = N_s[i] + N_n[i]
        
        # 3. Calculate the scaling coefficient gamma matrix (Equation 4)
        if E_sn[i] > 0:
            gamma[i] = N * (N_s[i] / E_sn[i])
        else:
            gamma[i] = 1.0 # Base multiplier if no events occurred in the bin
            
    # 4. Integrate across the arrays to derive final segment metrics (Equations 5 & 6)
    N_sv = np.sum((gamma * E_v) / N)
    N_sg = np.sum((gamma * E_g) / N)
    
    return N_sv, N_sg
```

```python
rho_v_over_rho_g_list = []
N_sv_list = []
N_sg_list = []
for i in range(0,len(df_seg)):
    # Step A: Aggregate photon vectors from along-track spatial data slices
    E_v = df_seg.profile_E_v.iloc[i]
    E_g = df_seg.profile_E_g.iloc[i]
    E_sn = df_seg.profile_E_sn.iloc[i]
    N_n = df_seg.profile_N_n.iloc[i]
    M = df_seg.profile_M.iloc[i]
    N = df_seg.profile_laser_shots_N.iloc[i]

    C_ALS = als_seg.als_canopy_cover.iloc[i]

    # Step B: Perform radiometric reconstruction to get true signal outputs
    N_sv, N_sg = apply_radiometric_correction(E_v, E_g, E_sn, N_n, M, N)
    N_sv_list.append(N_sv)
    N_sg_list.append(N_sg)
    # Using: rho_v / rho_g = (N_sv * (1 - C_ALS)) / (C_ALS * N_sg)
    if C_ALS > 0 and N_sg > 0:
        rho_v_over_rho_g = (N_sv * (1 - C_ALS)) / (C_ALS * N_sg)
        
        # Bound the solution to physically realistic limits (e.g., 0.1 to 2.0)
        # This prevents extreme spikes where the canopy completely masks out the ground profile (N_sg -> 0)
        rho_v_over_rho_g = np.clip(rho_v_over_rho_g, 0.1, 2.0)
    else:
        rho_v_over_rho_g = np.nan # Complete canopy blockage or open water gaps
    rho_v_over_rho_g_list.append(rho_v_over_rho_g)
```

```python
def df_to_las(df, x_col, y_col, z_col, output_name, 
              intensity_col=None, classification_col=None, epsg=None):
    """
    Converts a pandas DataFrame to a LAS file using laspy 2.x.
    """
    
    # 1. Create a new header
    # Point format 3 is a good default as it supports GPS time and color, 
    # but point format 0 or 1 works for basic X, Y, Z, Intensity.
    header = laspy.LasHeader(point_format=3, version="1.2")

    # 2. Handle EPSG / CRS
    if epsg:
        header.add_crs(epsg)

    # 3. Create the LasData object
    las = laspy.LasData(header)

    # 4. Assign primary coordinates
    las.x = df[x_col].values
    las.y = df[y_col].values
    las.z = df[z_col].values

    # 5. Handle Optional Fields
    if intensity_col and intensity_col in df.columns:
        las.intensity = df[intensity_col].values
    
    if classification_col and classification_col in df.columns:
        las.classification = df[classification_col].values

    # 6. Write to file
    las.write(output_name)
    print(f"Successfully wrote {len(df)} points to {output_name}")

    return las
```
